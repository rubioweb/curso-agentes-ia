```python
# Las importaciones
# agente, ejecutor, traza

from dotenv import load_dotenv # Carga las variables de entorno del archivo .env

# Agent: Clase para crear la identidad, instrucciones y herramientas de cada IA
# Runner: El motor que gestiona el flujo de la conversación y la ejecución de funciones
# trace: Decorador/herramienta para monitorizar y registrar los pasos internos del agente
from agents import Agent, Runner, trace

```

```python
# El punto de inicio habitual
# Busca el archivo .env en tu carpeta y activa las credenciales
load_dotenv(override=True)

import os

# Apaga el cliente de telemetría interno del SDK para evitar el error 401 de OpenAI
os.environ["OPENAI_TRACING_ENABLED"] = "false"
```

El SDK de OpenAI Agents, por defecto, viene configurado para atacar directamente 

a la API oficial de OpenAI usando el parámetro model="gpt-4o-mini". Como estoy usando Groq (con Llama),

 si ejecuto eta celda tal como está, me lanzara un error de autenticación.

Para que el SDK redirija sus peticiones a los servidores de Groq (o a Ollama en local),

 tenemos que pasarle un cliente personalizado compatible con la API de OpenAI

```python

# Crea un agente con nombre, instrucciones, modelo

#agent = Agent(name="Jokester", instructions="Eres un comico que cuentas chistes", model="gpt-4o-mini")

# Creo un agente con nombre, instrucciones y modelo usando Groq

#import os

# 1. TRUCO DE ENRUTAMIENTO: Engañamos al SDK configurando las variables globales.
# En lugar de usar un cliente explícito, redirigimos la dirección base hacia Groq.
os.environ["OPENAI_BASE_URL"] = "https://api.groq.com/openai/v1"

# Seteamos la clave de Groq dentro de la variable que el SDK busca por defecto.
os.environ["OPENAI_API_KEY"] = os.getenv("GROQ_API_KEY")

# 2. Creamos el agente de forma limpia utilizando la estructura nativa del SDK.
agent = Agent(
    name="Jokester", 
    instructions="Eres un cómico profesional que cuenta chistes cortos. Responde siempre en castellano.", 
    model="llama-3.3-70b-versatile" # Indicamos el modelo gratuito de Groq disponible
)
```

```python
# Muestra en pantalla la estructura interna del objeto 'agent' creado para verificar sus propiedades
agent
```

```python
# Ejecuta la broma con Runner.run(agent, prompt) y lurgo imprime final_output

# El bloque 'trace' monitoriza y muestra en consola los pasos internos que da el agente
with trace("Chiste sobre agentes autònomos de IA"):
    
    # Runner.run envía de forma asíncrona el mensaje al modelo y procesa la respuesta
    result = await Runner.run(agent, "Cuèntame un chiste sobre agentes autònomos de IA")
    
    # Imprime en pantalla únicamente el texto de la respuesta final devuelta por la IA
    print(result.final_output)
```
