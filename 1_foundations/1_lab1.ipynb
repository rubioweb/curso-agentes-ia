```python
# Revisa la clave — si no estás usando OpenAI, ¡revisa la clave que estés usando! Ollama no necesita clave.

import os
groq_api_key = os.getenv('GROQ_API_KEY')

if groq_api_key:
    print(f"La clave API de OpenAI existe y comienza por{groq_api_key[:8]}")
else:
    print("La clave API de OpenAI no está configurada; revisa la guía de solución de problemas en la carpeta setup")
    
```

```python
# Y ahora — la importantísima declaración de importación
# Si obtienes un error de importación — ve a la sección de solución de problemas en la carpeta Setup
# Incluso para otros proveedores de LLM como Gemini, todavía usas esta importación de OpenAI — consulta la Guía 9 para saber por qué
# Importa el cliente OpenAI para crear una instancia y poder realizar llamadas a la API de modelos de lenguaje.

from openai import OpenAI
```

```python
# Ahora creamos una instancia de la clase Openai
# Si no estás seguro de lo que significa crear una instancia de una clase - ¡ve a la carpeta de guías (guía 6)!
# Si obtienes un NameError - ve a la carpeta de guías (guía 6) para aprender sobre los NameError - siempre se pueden solucionar al instante
# Si no estás usando OpenAI, solo necesitas modificar esto ligeramente - las instrucciones precisas están en la guía de APIs de IA (guía 9)

openai = OpenAI(
    api_key=os.getenv('GROQ_API_KEY'),
    base_url="https://api.groq.com/openai/v1"
)
```

```python
# Crea una lista de mensajes en el formato de la familia OpenAI
# Cada elemento de la lista es un diccionario con: role: quién habla. 
#  Puede ser: user → el usuario,  assistant → el modelo,  system → instrucciones del sistema, content: el texto del mensaje.
# La preguna Cuanto es 2+2
messages = [{"role": "user", "content": "Cuanto es 2+2?"}]
```

```python
# La guía de APIs (guía 9) tiene instrucciones exactas para usar alternativas incluso más económicas o gratuitas a OpenAI
# Si obtienes un NameError, ve a la carpeta de guías (guía 6)
# Cambia el parámetro "model" por uno que Groq tenga en sus servidores de forma gratuita.

# Envía la conversación al modelo de IA seleccionado y genera una respuesta.
response = openai.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=messages
)
# Muestra en pantalla el contenido de la respuesta generada por el modelo.
print(response.choices[0].message.content)
```

```python
# # Y ahora - Le pido que cree una nueva pregunta:

question = "Por favor, propón una pregunta difícil y desafiante para evaluar el CI de alguien. Responde solo con la pregunta.."
messages = [{"role": "user", "content": question}]
```

```python
# Ahora el modelo creara la pregunta

response = openai.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=messages
)

question = response.choices[0].message.content

print(question)
```

```python
# # forma una nueva lista de mensajes

messages = [{"role": "user", "content": question}]
```

```python
# Pregúntalo de nuevo con la nueva pregunta

response = openai.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=messages
)

answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))

```

            Primero, pídale al programa de LLM que elija un área de negocio que valga la pena explorar para una oportunidad en IA agente.<br/>
            Luego, pídale al LLM que presente un problema específico de esa industria: algo desafiante que podría ser propicio para una solución de Agentic.<br/>
            Finalmente, tenemos una tercera convocatoria de LLM para proponer la solución de IA agencial. <br/>

```python
#El primer "agente" actúa como estratega (elige el sector).
#El segundo actúa como analista (encuentra el fallo).
#El tercero actúa como consultor (proporciona una solución).diseña el software.
# Crear el 1º mensaje:

messages = [{"role": "user", "content": "Elige un área de negocio que valga la pena explorar para una oportunidad de IA Agéntica. Responde solo con el nombre del área."}]

# Hacer la 1º llamada

response = openai.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=messages
)

# # Extraemos el texto de la respuesta de la IA (el sector elegido)
area_negocio = response.choices[0].message.content
print(f"Área de negocio elegida: {area_negocio}\n")
```

```python
# 2. Segunda llamada: Encontrar un problema 
messages = [{"role": "user", "content": f"En la industria de {area_negocio}, busca un problema específico que sea ideal para resolver con una solución de IA Agéntica. Sé conciso."}]

response = openai.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=messages
)
problema_detectado = response.choices[0].message.content
print(f"Problema detectado: {problema_detectado}\n")
```

```python
# 3. Tercera llamada: Proponer la solución
messages = [{"role": "user", "content": f"Basándote en el problema: '{problema_detectado}' en el sector de {area_negocio}, propone una solución detallada mediante un Agente de IA."}]

response = openai.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=messages
)
solution = response.choices[0].message.content

# Resultado final
#Markdown: Es un traductor. Toma texto con formato simple y lo convierte en código visual (títulos grandes, negritas, etc
#display: Es una función que muestra el resultado final.
from IPython.display import Markdown, display
display(Markdown(f"##  Propuesta de Agente de IA\n\n**Sector:** {area_negocio}\n\n**Problema:** {problema_detectado}\n\n---\n\n### Solución Propuesta:\n{solution}"))
```
