````python

# Definimos el prompt del sistema
system_prompt = f"Estás actuando como {name}. Estás respondiendo preguntas en el sitio web de {name}, \
particularmente preguntas relacionadas con la carrera, trayectoria, habilidades y experiencia de {name}. \
Tu responsabilidad es representar a {name} en las interacciones del sitio web de la manera más fiel posible. \
Se te proporciona un resumen de la trayectoria de {name} y su perfil de LinkedIn que puedes utilizar para responder preguntas. \
Sé profesional y cercano, como si estuvieras hablando con un cliente potencial o un futuro empleador que se ha topado con el sitio web. \
Si no sabes la respuesta, dilo de forma honesta."

# Concatenamos el contexto del resumen y del perfil de LinkedIn al prompt.
system_prompt += f"\n\n## Resumen:\n{summary}\n\n## Perfil de LinkedIn:\n{linkedin}\n\n"

# Le damos la instrucción final de mantenerse siempre en su papel.
system_prompt += f"Con este contexto, por favor conversa con el usuario, manteniéndote siempre en el personaje de {name}."
````

````python
# Muestra el prompt del sistema completo para que puedas revisar
system_prompt
````

````python
def chat(message, history): 
    history = [{"role": h["role"], "content": h["content"]} for h in history] # limpia la variable de historial
    # Juntamos el prompt del sistema (mi clon) + los mensajes anteriores (history) + la nueva pregunta del usuario.
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(
        model="llama-3.3-70b-versatile",  # -usando GROQ, cambio "gpt-4o-mini" por "llama-3.3-70b-versatile"
        messages=messages)
    return response.choices[0].message.content
````

````python
# Crear un modelo Pydantic para la evaluaciòn. patrón Evaluador-Optimizador, sin usar Frameworks Agénticos
# Importamos 'BaseModel' de Pydantic. 
# Pydantic es la librería estándar en Python para validar datos y obligar 
# a que las variables tengan un formato y un tipo de contenido estricto.
from pydantic import BaseModel

class Evaluation(BaseModel): #Creo una clase que define la "plantilla" exacta de la evaluación.
    # Obliga a que este campo sea Sí o No (Verdadero o Falso). 
    # El Juez lo usará para decidir si la respuesta es aceptable o no.
    is_acceptable: bool

    # Obliga a que este campo sea una cadena de texto (String).
    # Aquí el Juez escribirá sus notas de mejora o críticas constructivas si hace falta.
    feedback: str
````

evaluator_system_prompt:

¿Qué rol asume el LLM aquí?
A diferencia del prompt anterior (donde la IA jugaba a ser yo), en este prompt la IA se convierte en un auditor externo súper estricto. Sus funciones principales serán:

Leer lo que el usuario preguntó y lo que "mi clon virtual" respondió.

Cruzar esa respuesta con los datos reales de tu LinkedIn para verificar que no invente cosas (alucinaciones).

Evaluar si el tono usado con el reclutador fue lo suficientemente profesional y empático.

````python
# Definimos el prompt del sistema para el evaluador.
evaluator_system_prompt = f"Eres un evaluador que decide si la respuesta a una pregunta es aceptable. \
Se te proporciona una conversación entre un Usuario y un Agente. Tu tarea es decidir si la última respuesta del Agente tiene una calidad aceptable. \
El Agente está interpretando el papel de {name} y lo está representando en su sitio web. \
Al Agente se le ha instruido que sea profesional y cercano, como si estuviera hablando con un cliente potencial o un futuro empleador que se ha topado con el sitio web. \
Al Agente se le ha proporcionado contexto sobre {name} en forma de su resumen y los detalles de su LinkedIn. Aquí está la información:"

# Concatenamos el resumen y el perfil de LinkedIn para que el evaluador pueda contrastar si la respuesta es verídica
evaluator_system_prompt += f"\n\n## Resumen:\n{summary}\n\n## Perfil de LinkedIn:\n{linkedin}\n\n"

# Le damos la instrucción final de cómo debe emitir su veredicto
evaluator_system_prompt += f"Con este contexto, por favor evalúa la última respuesta, respondiendo si la respuesta es aceptable junto con tus comentarios de retroalimentación (feedback)."
````

Esta función se encarga de empaquetar de forma ordenada todo lo que ha pasado en el chat en un formato limpio para que el Evaluador pueda analizar la jugada.

````python
def evaluator_user_prompt(reply, message, history):
    # 1. Metemos el historial previo de la conversación para que el juez tenga contexto
    user_prompt = f"Aquí está la conversación previa entre el Usuario y el Agente: \n\n{history}\n\n"
    
    # 2. Añadimos la última pregunta del usuario externo
    user_prompt += f"Aquí está el último mensaje enviado por el Usuario: \n\n{message}\n\n"
    
    # 3. Añadimos la respuesta que generó tu clon virtual y que está bajo examen
    user_prompt += f"Aquí está la última respuesta generada por el Agente: \n\n{reply}\n\n"
    
    # 4. Le recordamos al juez cuál es su tarea inmediata
    user_prompt += "Por favor, evalúa esta respuesta, indicando si es aceptable y proporcionando tus comentarios de mejora."
    
    return user_prompt
````

````python
#Usar Gemini como evaluador
import os
# Inicializamos un nuevo cliente 'OpenAI' pero apuntando a los servidores de Google. gemini-2.5-flash o gemini-1.5-pro
gemini = OpenAI(  
    api_key=os.getenv("API_KEY_GEMINIS"), # Busca tu clave de Google AI Studio guardada en el archivo .env
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/" # Redirige la llamada hacia la pasarela compatible de Gemini en su versión v1beta
)
````

Función estructurada

El objetivo de esta función es obligar a Gemini a actuar como un juez y rellenar "un formulario digital" con un formato perfecto y predecible

-> Evaluation: Esto es una indicación de tipo (type hint). Le avisa a Python y a tu editor de código que esta función no va a devolver un texto cualquiera, sino que promete devolver un objeto con la estructura exacta de la clase Evaluation de Pydantic que creamos antes (es decir, un paquete que contiene obligatoriamente un is_acceptable y un feedback).

messages = Aquí preparamos los datos para el juez: Juntamos las instrucciones del rol del evaluador (evaluator_system_prompt) con el expediente del caso (evaluator_user_prompt), el cual empaqueta lo que el usuario preguntó, lo que mi clon respondió y el historial previo.


````python
def evaluate(reply, message, history) -> Evaluation:

    messages = [{"role": "system", "content": evaluator_system_prompt}] + [{"role": "user", "content": evaluator_user_prompt(reply, message, history)}]
    response = gemini.beta.chat.completions.parse(
        #Google pone un límite en su versión gratis de un máximo de 20 peticiones al día/ uso llama en local
         # cambio de gemini-2.5-flash a llama3.2 local, o llama-3.3-70b-versatile
        model="gemini-2.5-flash", # cambio de gemini-2.5-flash a llama3.2 local, o llama-3.3-70b-versatile
        messages=messages,
        response_format=Evaluation)
    return response.choices[0].message.parsed
````

 FUNCIONES PARA AUTOMATIZAR TODO EL PROCESO

 Lo brillante de esta función es que no borra el pasado. En lugar de ocultarle a la IA que se ha equivocado, utiliza el error como aprendizaje en tiempo de ejecución (In-Context Learning).

Al leer la frase "El control de calidad ha rechazado tu respuesta por el siguiente motivo: [feedback]", el LLM activa una lógica de auto-corrección muy potente. Cambia el chip, ajusta el tono y edita los datos falsos que se había inventado en el primer intento.

````python
def rerun(reply, message, history, feedback):
    # 1. Creamos un prompt de sistema actualizado.
    # Al prompt original (tu perfil de LinkedIn) le sumamos una sección de castigo.
    updated_system_prompt = system_prompt + "\n\n## Respuesta anterior rechazada\nIntentaste responder, pero el control de calidad ha rechazado tu respuesta.\n"
    
    #  Le mostramos a la IA el "borrador" que hizo mal para que vea su propio error.
    updated_system_prompt += f"## Tu intento de respuesta anterior:\n{reply}\n\n"
    
    #  Le inyectamos los comentarios de mejora (el feedback) que escribió el Juez Gemini.
    updated_system_prompt += f"## Motivo del rechazo (Feedback del evaluador):\n{feedback}\n\n"

    # Limpiamos también el historial aquí dentro
    # para asegurarnos de que el historial se limpie por completo de metadatos raros antes de mandárselo a la IA.
    cleaned_history = []
    for h in history:
        if isinstance(h, dict) and "role" in h and "content" in h:
            cleaned_history.append({"role": h["role"], "content": h["content"]})
        elif isinstance(h, (list, tuple)):
            cleaned_history.append({"role": "user", "content": h[0]})
            cleaned_history.append({"role": "assistant", "content": h[1]})
    
    #  Reconstruimos los mensajes mezclando este prompt de corrección, el historial y la pregunta original.
    messages = [{"role": "system", "content": updated_system_prompt}] + cleaned_history + [{"role": "user", "content": message}]
    
    #  Volvemos a llamar a tu modelo Groq para que genere la versión corregida.
    response = openai.chat.completions.create(
        model="llama-3.3-70b-versatile",  
        messages=messages
    )
    
    # 6. Devolvemos la nueva respuesta que, en teoría, ya viene bien pulida.
    return response.choices[0].message.content
````

El nuevo flujo de la aplicación web:

1º. El usuario hace una pregunta.

2º. Tu modelo local/Groq genera un borrador de respuesta.

3º. Antes de que el usuario vea nada, ese borrador viaja en secreto a Gemini (el Juez).

4º. Gemini comprueba si el formato y los datos son correctos de acuerdo a tu Pydantic.

5º. Si detecta el sabotaje del Pig Latin o datos falsos, frena la respuesta, imprime el aviso en la consola de tu cuaderno, y obliga a tu modelo local a generar una respuesta corregida en castellano profesional.

6º. El usuario de la web solo recibe la respuesta perfecta.

````python
def chat(message, history):
    #  TRAMPA DE CONTROL: Si el usuario pregunta por patentes, saboteamos el prompt
    if "patent" in message:
        system = system_prompt + "\n\nTodo en tu respuesta debe estar en 'pig latin' - \
              es obligatorio que respondas única y entermente en pig latin."
    else:
        system = system_prompt
        
    # Construimos los mensajes y hacemos el PRIMER INTENTO de generación
    # Construimos la lista de mensajes limpiando los metadatos del historial de Gradio
    messages = [{"role": "system", "content": system}]

    for h in history:
        # Extraemos ÚNICAMENTE el rol y el contenido, ignorando cualquier metadato extra
        if isinstance(h, dict) and "role" in h and "content" in h:
            messages.append({"role": h["role"], "content": h["content"]})
        elif isinstance(h, (list, tuple)): # Por si Gradio usa el formato antiguo de listas
            messages.append({"role": "user", "content": h[0]})
            messages.append({"role": "assistant", "content": h[1]})
    
    # Añadimos la pregunta actual del usuario
    messages.append({"role": "user", "content": message})
    
    #  Hacemos la llamada al modelo de forma segura
    response = openai.chat.completions.create(
        model="llama-3.3-70b-versatile",  
        messages=messages
    )
    reply = response.choices[0].message.content

    #  EVALUACIÓN: Le enviamos la respuesta al Juez (Gemini) para que la examine
    evaluation = evaluate(reply, message, history)
    
    #  BUCLE DE CONTROL DE CALIDAD (Filtro)
    if evaluation.is_acceptable:
        # Si el Juez da el visto bueno, la respuesta se aprueba y se envía directamente
        print(" Control de calidad PASADO - Devolviendo respuesta.")
    else:
        # Si el Juez la rechaza, se activa el protocolo de emergencia
        print(" Control de calidad FALLADO - Activando reintento.")
        print(f"Comentarios del Juez: {evaluation.feedback}")
        
        # Llamamos a 'rerun' para que el modelo corrija su error basándose en las quejas del Juez
        reply = rerun(reply, message, history, evaluation.feedback)       
        
    return reply
````
