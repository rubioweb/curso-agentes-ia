````python
# Empieza por las importaciones: pídele a ChatGPT que te explique cualquier paquete que no conozcas.

import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from anthropic import Anthropic
from IPython.display import Markdown, display
````

````python
# ¡Recuerda siempre hacer esto!
load_dotenv(override=True)
````

````python
# Imprime los prefijos de las claves para facilitar la depuración.

openai_api_key = os.getenv('OPENAI_API_KEY')
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
google_api_key = os.getenv('API_KEY_GEMINIS')
deepseek_api_key = os.getenv('DEEPSEEK_API_KEY')
groq_api_key = os.getenv('GROQ_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("No se ha configurado la clave de API de OpenAI")
    
if anthropic_api_key:
    print(f"Anthropic API Key exists and begins {anthropic_api_key[:7]}")
else:
    print("No se ha configurado la clave de API de Anthropic (y esto es opcional)")

if google_api_key:
    print(f"Google API Key exists and begins {google_api_key[:2]}")
else:
    print("No se ha configurado la clave de API de Google (y esto es opcional)")

if deepseek_api_key:
    print(f"DeepSeek API Key exists and begins {deepseek_api_key[:3]}")
else:
    print("No se ha configurado la clave de API de DeepSeek (y esto es opcional)")

if groq_api_key:
    print(f"Groq API Key exists and begins {groq_api_key[:4]}")
else:
    print("No se ha configurado la clave de API de Groq (y esto es opcional)")
````

````python
# Realizo la pregunta concatenada

request = "Por favor, propón una pregunta desafiante y matizada que pueda hacerle a varios LLM para evaluar su inteligencia. "
request += "Responde solo con la pregunta, nada más."
messages = [{"role": "user", "content": request}]
````

````python
# muestro la pregunta
messages
````

````python
# PARA QUE LA IA PROPONGA UNA PREGUNTA DESAFIANTE

# Configuramos la instancia global para que apunte a Groq usando mi clave
# 1. Creamos y configuramos el objeto 'openai' para conectarnos a Groq.
#Aunque la librería se llama OpenAI, le pasamos la clave y la dirección (base_url) de Groq
#para "redirigir" todas nuestras peticiones hacia sus servidores gratuitos.
openai = OpenAI(
    api_key=os.getenv('GROQ_API_KEY'),
    base_url="https://api.groq.com/openai/v1" ## Le dice a la librería que viaje a Groq en vez de a ChatGPT
)

# 2. Enviamos la petición al servidor de Groq para generar la respuesta.
# Usamos el método 'chat.completions.create' que es el estándar para pedir respuestas de texto.
response = openai.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=messages,  # Le pasamos la  pregunta
)

# 3. Extraemos el texto limpio de la respuesta.
# Con esta línea entramos a la primera opción '[0]' y extraemos solo el contenido del mensaje de texto.
question = response.choices[0].message.content
print(question)
````

````python
#Crea una lista vacía donde se irán guardando los nombres de las inteligencias artificiales que van a competi
# Crea otra lista vacía donde guardaremos la respuesta
#Guarda la pregunta súper compleja del robot y los obstáculos gaussianos que acabamos de generar dentro del formato correcto,

competitors = []
answers = []
messages = [{"role": "user", "content": question}]
````

````python
# El primer competidor (Groq - Llama)
# En la celda que originalmente era para gpt-5-nano, utilizo  el modelo gratuito de Groq (llama-3.1-70b-versatile).
model_name = "llama-3.3-70b-versatile"

response = openai.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
````

````python
# Anthropic has a slightly different API, and Max Tokens is required
# api de pago
model_name = "claude-sonnet-4-5"

claude = Anthropic()
response = claude.messages.create(model=model_name, messages=messages, max_tokens=1000)
answer = response.content[0].text

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
````

````python
# # Segundo competidor: Gemini de Google usando mi clave

gemini = OpenAI(api_key=google_api_key, base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
model_name = "gemini-2.5-flash"

response = gemini.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
````

````python
# APIS DE PAGO
deepseek = OpenAI(api_key=deepseek_api_key, base_url="https://api.deepseek.com/v1")
model_name = "deepseek-chat"

response = deepseek.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
````

````python
#  Actualizado con el último modelo de código abierto de OpenAI
# La plataforma te da acceso a modelos de última generación, 
# como los nuevos gpt-oss de OpenAI, además de las familias Llama, Gemma y Qwen
# openai/gpt-oss-120b: Modelo masivo de OpenAI (117B params), ideal para razonamiento complejo. Velocidad: 500+ tokens/s
groq = OpenAI(api_key=groq_api_key, base_url="https://api.groq.com/openai/v1")
model_name = "openai/gpt-oss-120b"

response = groq.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
````

````python
ollama = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
model_name = "llama3.2"

response = ollama.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
````

````python
# ¿Dónde estamos?

print(competitors)
print(answers)
````

````python
# uso de "zip" para iterar sobre dos listas simultaneamente
#La función zip funciona como una cremallera. Va uniendo el primer elemento
#  de la lista A competitors con el primer elemento de la lista B, luego el segundo con el segundo (answers)., y así sucesivamente.
for competitor, answer in zip(competitors, answers):
    print(f"Competitor: {competitor}\n\n{answer}")
````

````python
# vamos a juntarlo todo. ver uso de "enumerate"
# Creo una variable de texto (String) vacía.donde ire "pegando" todas las respuestas una detrás de otra.
together = ""
# Iniciamos un bucle usando 'enumerate'.
# A diferencia de un 'for' normal, 'enumerate' nos da dos cosas a la vez en cada vuelta:
#   - El 'index': El número de posición (0, 1, 2...)
#   - El 'answer': El texto de la respuesta en sí.
for index, answer in enumerate(answers):
    # Pegamos un título para saber de quién es la respuesta.
    # Como los índices en programación empiezan en 0, le sumamos '+1' 
    # para que en la pantalla quede bonito como "competitor 1", "competitor 2", etc.
    together += f"# Respuesta del competidor {index+1}\n\n"
    # Sumamos el contenido de la respuesta y dejamos un par de saltos de línea (\n\n)
    # para que no quede todo el texto apelotonado y sea fácil de leer.
    together += answer + "\n\n"
````

````python
print(together)
````

````python
# PLANTILLA PARA EL JUEZ
# Definimos la plantilla de texto (f-string) que servirá de instrucción para la IA que actúe como juez.
judge = f"""Actúa como un juez severo en una competición entre {len(competitors)} modelos de Inteligencia Artificial.
A cada modelo se le ha planteado la siguiente pregunta original:

{question}

Tu única tarea es evaluar cada una de las respuestas recibidas analizando su claridad, la fuerza de sus argumentos y su coherencia. 
Después, debes clasificarlas en orden de mejor a peor.

Es un requisito OBLIGATORIO que respondas ÚNICAMENTE con un formato JSON estructurado de la siguiente manera, sin añadir texto extra:
{{"results": ["número del mejor competidor", "número del segundo mejor competidor", "número del tercer mejor competidor", ... ]}}

Aquí tienes las respuestas acumuladas de cada competidor para que las analices:

{together}

Ahora, devuelve exclusivamente el JSON con el orden de clasificación de los competidores. No añadas introducciones, saludos ni bloques de código de markdown (```json). Devuelve solo el texto plano del JSON."""
````

````python
# Imprime por pantalla todo el texto del prompt que configuramos antes.
print(judge)
````

````python
# Empaqueta el texto del juez dentro de la estructura oficial de mensajes de la API.
# Las IA no entienden el texto suelto; necesitan que les digamos quién está hablando.
# Aquí le indicamos que el rol es "user" (el usuario que hace la petición) y que el contenido es el prompt 'judge'.
judge_messages = [{"role": "user", "content": judge}]
````

````python
# Judgement time!
#openai = OpenAI()
#response = openai.chat.completions.create(
 #   model="gpt-5-mini",
  #  messages=judge_messages,
#)
#results = response.choices[0].message.content
#print(results)

#Usar  Ollama local (llama3.2) para la respuesta
# Conecto con mi Ollama local
ollama_juez = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')

# Le enviamos el caso a Llama 3.2
response = ollama_juez.chat.completions.create(
    model="llama3.2",
    messages=judge_messages,
)

results = response.choices[0].message.content
print(results)
````

````python
# Corrección manual para que el cuaderno entienda el JSON de Llama 3.2
results = '{"results": [1, 2, 3]}'
print("JSON corregido para el cuaderno:", results)
````

````python
# OK let's turn this into results!

results_dict = json.loads(results)
ranks = results_dict["results"]
for index, result in enumerate(ranks):
    competitor = competitors[int(result)-1]
    print(f"Rank {index+1}: {competitor}")
````
